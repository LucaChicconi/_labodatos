### 4. Empezando con _penguins_ escribir una secuencia de operaciones que :

(a) Excluya a los pingüinos observados en la isla Biscoe.

(b) Sólo se quede con las variables que están entre _species_ y *body_mass_g* inclusive.

(c) Renombre la variable _species_ a *especie_pinguino*

(d) Agrupe los datos por la variable *especie_pinguino*

(e) Calcule el valor medio de las variables que contienen el valor "length", separado por la especie del pingüino, y llamando a las columnas como las originales pero agregando el "_mean" al final.

````python
# (a)
penguinsNotBiscoe = penguins[penguins["island"]!="Biscoe"]
# (b)
penguinsNotBiscoe = penguinsNotBiscoe[[col for col in penguins.columns if not(col in ['sex'])]]
# (c)
penguinsNotBiscoe.rename(columns= {"species":"especie_pinguino"},inplace = True)


# (d)

penguinsNotBiscoe.groupby("especie_pinguino")

# (e)

penguinsNotBiscoe.groupby("especie_pinguino").agg({col: "mean" for col in penguinsNotBiscoe.columns if "length" in col}).rename(columns={"bill_length_mm":"bill_length_mm_mean","flipper_length_mm":"flipper_length_mm_mean"})



````
